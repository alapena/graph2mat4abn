```python
# Just to define file paths, not related to sisl
from pathlib import Path
import numpy as np

# Sisl imports
import sisl
import sisl.viz

# To quickly plot the hamiltonian matrix
import plotly.express as px
```

```python
#Get the fdf file for the calculation
path_uc = Path("../dataset/SHARE_OUTPUTS_8_ATOMS/0a2a-fbca-4649-8012-f5aa640bfd1d")   #SIESTA calculation folder
fdf_uc = sisl.get_sile(path_uc / "aiida.fdf")        #the fdf file inside it (the name xxxx.fdf could change)
#Read the Hamiltonian data
H_uc = fdf_uc.read_hamiltonian()

geometry = fdf_uc.read_geometry()
cell = geometry.cell
```

```python
kpt = np.array([1,8,1], dtype=np.float64)

# Me:
from graph2mat4abn.tools.tools import reconstruct_tim_from_coo, reduced_coord
H_me = reconstruct_tim_from_coo(kpt, H_uc.tocsr().tocoo(), geometry, cell)

# Sisl:
H_sisl = H_uc.Hk(reduced_coord(kpt, cell))
# px.imshow(np.abs(H_me))

#check they are the same
MatsClose=np.allclose(H_me, H_sisl.toarray())
print(f'np.allcose between the two transfer integral matrices returns: {MatsClose}')
```
